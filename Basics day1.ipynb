Type Conversion

> Implicit vs explicit

```python
# Implicit
print(5+5.6)
print(type(5+5.6))
print(type(5),type(5.6))

# print(4+'4')   error
```

```python
# explicit

# str->int
print(type(int('4')))
# print(type(int('d')))  give error

# float->int
print(type(int(4.5)))

# int->float
print(type(float(4)))

# int->str
print(str(75))
print(type(str(75)))

# complex->int (error)
print(4+6+5j)
print(type(4+6+5j))

# print(int(4+6+5j))   information lose it will give error
```

Variables

>Variables are named storage locations (containers) in memory that hold data values, which can be changed or varied during the execution of a program.

Identifier

>An identifier is a user-defined name used to uniquely identify entities such as variables, functions, classes, or structures within a program

Literals
>  literals are fixed, hard-coded values written directly into the source code, such as numbers, strings, or characters, which represent themselves and cannot be altered once defined.

(varibale ko jab bhi value milta ha to use literals kahte hai)




```python
# Literals

a=0b1010 #Binary Literals(0b)
print(a)

b=100  #Decimal Literal
print(b)

c=0o310 #Octal Literal(0o)
print(c)

d=0x12c #Hexadecimal Literal(0x)
print(d)

# Float Literal
float_1=10.5
float_2=1.5e2 #1.5*10^2
float_3=1.5e-3 #1.5*10^-3

print(float_1,float_2,float_3)

# Complex Literal
x=3.14j
print(x,x.imag,x.real)

# string
string='This is python'
strings="This is python"
char="C"
multiline_str="""This is a multiline string with more than one line code."""
unicode=u"\U0001f600\U0001F606\U0001F923"
raw_str=r"raw \n \t string"

print(string)
print(strings)
print(char)
print(multiline_str)
print(unicode)
print(raw_str)

a=True+4
b=False+10

print("a:",a)
print("b:",b)

a=None      #helpful to declare variable in advance
print(a)
```
