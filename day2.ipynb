````python
#Arithmetic Operators
print(5+6)
print(5-6)
print(5*6)
print(5/6)
print(5//2)  #integer Division
print(5%2)   #Modulus
print(5**2)  #Exponent

````

````python
#Relation operator
print(4>5)
print(4<5)
print(4>=4)
print(4<=4)
print(4==4)
print(4!=4)
````

````python
#Logical operator
print(1 and 0)
print(1 or 0)
print(not 1)
````

````python
#Bitwise operator

# bitwise and
print(2&3)

# bitwise or
print(2|3)

#bitwise xor
print(2^3)

#bitwise not
print(~3)

#bitwise left shift
print(4>>2)

#bitwise right shift
print(4<<2)
````

````python
#Assignment operator

#=
#a=2

a=2

# a=a+2
a+=2

print(a)

````

````python
#Membership operator

#in / not in

print('D' in 'Delhi')

print('D' not in 'Delhi')

print(1 in [2,4,3,5,1])
````

````python
#guessing game
import random

jackpot=random.randint(1,100)
guess=int(input("Guess a number between 1 to 100"))
counter=1

while guess !=jackpot:
  if(guess<jackpot):
   print('galat! guess higher')
  else:
    print('galat! guess lower')
  guess=int(input("Guess the number again"))
  counter+=1
else:
  print('correct guess')
  print('attempts',counter)

````
