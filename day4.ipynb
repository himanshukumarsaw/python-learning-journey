# Tuples

A tuple in Python is similar to a list. The difference between the two is that we cannot change the elements of a tuple once it is assigned whereas we can change the elements of a list.

In short, a tuple is an immutable list. A tuple can not be changed in any way once it is created.

Characterstics

- Ordered
- Unchangeble
- Allows duplicate

````python
L=(43,54,65,43)
L1=(54,43,65,43)
print(L==L1)
# L[4]=3      'tuple' object does not support item assignment
print(L)
````

# Difference between List and Tuples


*   Syntax
*   Mutability
*   Speed  (tuples are fast immutable data types are faster)
*   Memory  (list consume more space,same above reason)
*   Built in functionality (List provide more function)
*   Error prone (List are More Error prone)
*   usability





````python
import time
L=list(range(100000000))
T=tuple(range(100000000))

start=time.time()
for i in L:
  i*5
print('List time',time.time()-start)

start=time.time()
for i in T:
  i*5
print('Tuple time',time.time()-start)
````

````python
import sys
L=list(range(1000))
T=tuple(range(1000))

print('List size',sys.getsizeof(L))
print('Tuple size',sys.getsizeof(T))
````

# Sets

A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc.

Characterstics:
- Unordered
- Mutable
- No Duplicates
- Can't contain mutable data types

### Creating Sets

````python
#empty
s={}             #it not create a set . it create dictionary
print(type(s))

s=set()
print(s)
print(type(s))

#1D and 2D
s1={1,2,3}
s2={1,2,3,(4,5,6)}
# s2={1,2,3,{4,5,6}}     unhashable type: 'set'  because it not contain mutable element
print(s1)
print(s2)

#Homogeneous and heterogeneous
s3={1,'hello',4.5,True,(1,2,3)}
print(s3)

#USING type conversion
s4=set([1,2,3])
print(s4)

#duplicates not allowed
s5={ 1,3,54,2,4,5,5,4}
print(s5)

#set can't have mutable items
# s6={1,2,[3,4]}
# print([s6])
````

#Adding items


*   Add

````python
s={1,3,4}
#add
s.add(5)   #add can add only one element
s.add(34)
print(s)
#update
s.update([5,6,7])   #update can add multiple element using list
s.update([89])
print(s)
````

Deleting Items


````python
# del
s={1,2,3,53,23,3,34,50}
print(s)
# del s[0]   'set' object doesn't support item deletion

#discard
s.discard(50)      #if the element is present then remove it
s.discard(90)    #if it not present it will not give error
print(s)

#remove
s.remove(3)
# s.remove(75)  it will give error when there is no element found
print(s)

#pop
s.pop()           #it will delete randomly
print(s)

s.clear()        #it will clear all element =>set()  [Empty set]
print(s)
````

Set Operation

````python
s1={1,2,3,4,5}
s2={4,5,6,7,8}
#Union(|)
print(s1|s2)
#intersection(&)
print(s1&s2)
#difference(-)
print(s1-s2)
print(s2-s1)
#symmetric Difference(^)
print(s1^s2)
#Membership Test
print(1 not in s1)
#Iteration
for i in s1:
   print(i)
````

````python
#len/sum/min/max/sorted
s={3,1,4,52,7}
print(len(s))
print(sum(s))
print(min(s))
print(max(s))
print(sorted(s,reverse=True))

````

````python
#intersection/intersection_update
s1={1.2,3,4,5}
s2={4,5,6,7,8}

print(s1.intersection(s2))
s1.intersection_update(s2)

print(s1)
print(s2)
````

````python
#difference/difference_update
s1={1,2,3,4,5}
s2={4,5,6,7,8}

print(s1.difference(s2))
s1.difference_update(s2)
print(s1)
print(s2)
````

````python
#Symmetric_difference/Symmetric_difference_update
s1={1,2,3,4,5}
s2={4,5,6,7,8}

s1.symmetric_difference_update(s2)

print(s1.symmetric_difference(s2))

print(s1)
print(s2)
````

# Dictionary

Dictionary in Python is a collection of keys values, used to store data values like a map, which, unlike other data types which hold only a single value as an element.

In some languages it is known as map or assosiative arrays.

dict = { 'name' : 'nitish' , 'age' : 33 , 'gender' : 'male' }

Characterstics:

- Mutable
- Indexing has no meaning
- keys can't be duplicated
- keys can't be mutable items

````python
#EMPTY DICTIONARY
d={}
print(d)
print(type(d))

#1D dictionary
d1={'name':'nitish','gender':'male'}
print(d1)

# with mixed keys
d2={(1,2,3):1,'hello':'world'}
print(d2)

#2D dictionary
s={
    'name':'himanshu',
    'college':'sviet',
    'sem':4,
    'subjects':{
        'dsa':50,
        'maths':67,
        'english':34
    }

}
print(s)


#using sequence and dict function
d4=dict([('name','nitish'),('age',32),(1,1)])
print(d4)

# duplicate keys  (not allowed -> if there is duplicate key it will opt to latest one)
d5={'name':'himanshu','name':'aklavya'}
print(d5)


#mutable items as keys
# d6={'name':'nitish',[1,2,3]:2}  unhashable type
# print(d6)

d6={'name':'nitish',(1,2,3):2}
print(d6)
````

# Remove key-value pair
1.   pop
2.   popitem
3.   del
4.   clear





````python
d={'name':'nitish','age':32,'gender':'male',3:3,'weight':43.2}

#pop               It delete the specific key
print(d.pop('name'))
print(d)

#popitem           It delete the item from back
d.popitem()
print(d)

#del               It delete the item by key
del d['gender']
print(d)

#clear
d.clear()
print(d)
````

# dictionary Comprehension

````python
#print 1st 10 numbers and their squares
{i:i**2 for i in range(1,11)}
````

````python
distances={'delhi':1000,'mumbai':2000,'banglore':3000}
print(distances.items())
````

````python
#using existing dict
distances={'delhi':1000,'mumbai':2000,'banglore':3000}
{key:value*0.62 for (key,value) in distances.items()}
````

````python
days=["sunday","Monday","Tuesday","wednesday","Thrusday","Friday","saturday"]
temp_c=[30.5,32.6,31.8,33.4,29.8,30.2,29.9]
print(zip(days,temp_c))
````

````python
#using zip
days=["sunday","Monday","Tuesday","wednesday","Thrusday","Friday","saturday"]
temp_c=[30.5,32.6,31.8,33.4,29.8,30.2,29.9]


{i:j for (i,j) in zip(days,temp_c)}
````

````python
#using if condition
products={'phone':10,'laptop':0,'charger':32,'tablet':0}

{key:value for (key,value) in products.items() if value>0}
````

````python
#nested Comprehension
#print tables of number from 2 to 4
{i:{j:i*j for j in range(1,11)} for i in range(2,5)}
````
