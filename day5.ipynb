#Types of Argument

*   Default Argument
*   Positional Argument
*   keyword Argument



````python
def power(a,b):
  return a**b

# power(2)    power() missing 1 required positional argument: 'b'


#default argument
def power(a=1,b=1):
  return a**b
print(power(2) )


#positional argument
def power(b=1,a=1):
  return b
print(power(4,5))

#keyword argument
power(b=3,a=3)

power(5,a=4)
````

*args and **kwargs


```
*args and **kwargs are special python keywords that are used to pass the variable length of
arguments to a function.
```



````python
# *args
# allows us to pass a variable number of non-keyword arguments to a function.
def multiply(a,b,c):
  return a*b*c
print(multiply(3,5,3))

#-----------------------------------------------------

def multiply(*args):
  product=1

  for i in args:
    product=product*i
  print(args)
  return product
print(multiply(1,3,4,6,8))

# ----------------------------------------------------

def multiply(a,b,c,*args):
  product=1

  for i in args:
    product=product*i
  print(args)
  return product
print(multiply(1,3,4,6,8))

# ----------------------------------------------------

#Too access any documentation
print(is_even.__doc__)
print(type.__doc__)
print(print.__doc__)
````

````python
# **kwargs
# **kwargs allow us to pass any number of keyword arguments.
#keyword arguments means that they contain a key-value pair,like a python dictionary

def display(**kwargs):
  print(kwargs)
  print(kwargs.items())
  print(type(kwargs))
  for (key,value) in kwargs.items():
    print(key,'->',value)
display(india='delhi',sriLanka='colombo',pakistan='islamabad')
````

points to remember while using *args and **kwargs

*   order of the arguments matter(normal->*args->**kwargs)
*   The words "args" and "kwargs" are only a conversation,you can use any name of your choice



````python
def display(a,b,c,*args,**kwargs):
  print(a)
  print(b)
  print(c)
  for i in args:
    print(i)
  for (key,value) in kwargs.items():
    print(key,'->',value)
display(2,4,7,3,64,3,india='delhi',sriLanka='colombo',pakistan='islamabad')
````

#Lambda Function
A lambda function is a small anonymous function.

A lambda function can take any number of arguments,but can only have one expression.

lambda a,b:a+b

Lambda keyword:creates the lambda expression

parameters: One or more parameters are supported.Must be separated by a comma(,) and no parantheses.

colon: This is a cue for the expression

Expression:Must be a single valid Python expression


````python
#x->x^2
a=lambda x:x**2
print(a)
a(2)
````

````python
#x,y->x+y
a=lambda x,y:x+y
a(5,2)
````

#diff between lambda vs Normal function

*   No name
*   lambda has no return value(infact,returns a function)
*   lambda is written in 1 line
*   not reusable

Then why use lambda function?

They are used with HOF


````python
#check if a string has 'a'
a=lambda s:'a' in s
a('hello')
````

````python
#odd or even
a=lambda x:'even' if x%2==0 else 'odd'
a(6)
````

#Higher order function
hof usko bolte hai jisme ham koi function ke andar dusre function return karte hai

 ya fir ham kisi function ke andar other function input as parameter pass karte hai

````python
#where we need Hof
def square(x):
  return x**2

def cube(x):
  return x**3


def transform(f,L):
  output=[]
  for i in L:
    output.append(f(i))
  print(output)

L=[1,2,3,4,5]
transform(square,L)
transform(cube,L)
````
