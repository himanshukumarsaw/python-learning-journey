````python
# readline()-> to read line by line
f=open("sample.txt",'r')
print(f.readline(),end='')
print(f.readline(),end='')
f.close()
````

````python
f=open("sample.txt",'r')
print(f.readline())
print(f.readline())
f.close()
````

````python
# reading entire using readline

f=open('sample.txt','r')

while True:
  data=f.readline()

  if data=="":
    break
  else:
    print(data,end="")
f.close()
````

````python
# readlines()-> to

# readline()-> to read line by line
f=open("sample.txt",'r')
s=f.readlines()
print(s)
for i in s:
  print(i,end="")
f.close()
````

# using context manager(with)


*   it's a good idea to close a file after usage as it will free up the resources
*   if we don't close it,garbage collector would close it
*   with keyword closes the file as soon as the usage is over



````python
# with
with open('hs.txt','w') as f:
  f.write('Himanshu gupta')
````

````python
# f.write("sonam gupta")    I/O operation on closed file.
````

````python
# try f.read() now
with open("hs.txt",'r') as f:
  print(f.read())
````

````python
#moving within a file -> 10 char then 10 char
with open('sample.txt','r') as f:
  print(f.read(2))
  print(f.read(3))

#buffer every time track karte rahta hai ki abhi
#  tak kitne char ko process kiya hai or hamesha uska count rakta hai
#to next time jab read karega to wo wahi se read karega
````

````python
#benefits ? -> to load a big file in memory
big_L=['hello world' for i in range(1000)]

with open('big.txt','w') as f:
  f.writelines(big_L)
````

````python
with open('big.txt','r') as f:
  chunk_size=100

  while len(f.read(chunk_size))>0:
    print(f.read(chunk_size),end="")
    f.read(chunk_size)
````

````python
#seek and tell function
with open('sample.txt','r') as f:
  print(f.read(2))
  print(f.tell())
  print(f.read(4))
  f.seek(0)
  print(f.read(4))
````

````python
# seek during write

with open('sample.txt','w') as f:
  f.write('Hello')
  f.seek(0)
  f.write('Xa')
````
